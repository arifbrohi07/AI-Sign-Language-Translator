# AI Sign Language Translator - Model Training & Evaluation

This notebook trains and evaluates machine learning models (Random Forest, SVM, MLP) for classifying ASL hand gestures using normalized 21-point hand landmark coordinates.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
import joblib
```

```python
# Load Landmark Dataset
df = pd.read_csv('../data/landmarks.csv')
print(f"Dataset shape: {df.shape}")
print(f"Target distribution:\n{df['label'].value_counts()}")
```

```python
# Train/Test Split (80% Train, 20% Test Stratified)
X = df.drop(columns=['label']).values
y = df['label'].values
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
print(f"X_train: {X_train.shape}, X_test: {X_test.shape}")
```

```python
# Train Random Forest Classifier
rf = RandomForestClassifier(n_estimators=100, max_depth=16, random_state=42, n_jobs=-1)
rf.fit(X_train, y_train)
y_pred = rf.predict(X_test)

acc = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {acc * 100:.2f}%")
print(classification_report(y_test, y_pred))
```

```python
# Plot Confusion Matrix
labels = sorted(list(set(y_test)))
cm = confusion_matrix(y_test, y_pred, labels=labels)
plt.figure(figsize=(14, 12))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.title('Sign Language Classification Confusion Matrix')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.show()
```

```python
# Export Model Artifacts
joblib.dump(rf, '../models/sign_language_model.pkl')
print('Saved model to ../models/sign_language_model.pkl')
```
